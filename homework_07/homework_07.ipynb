# I. Numpy

### Импортируйте NumPy

### Создайте одномерный массив размера 10, заполненный нулями и пятым элемент равным 1. Трансформируйте в двумерный массив.

```python
arr = np.zeros(10)
arr[4] = 1
arr_2d = arr.reshape(2, 5)
arr_2d
```

### Создайте одномерный массив со значениями от 10 до 49 и разверните его (первый элемент становится последним). Найдите в нем все четные элементы.

```python
arr = np.arange(10, 50)
arr_reversed = arr[::-1]
even_elements = arr_reversed[arr_reversed % 2 == 0]
print("Развёрнутый массив:", arr_reversed)
print("Чётные элементы:", even_elements)
```

### Создайте двумерный массив 3x3 со значениями от 0 до 8

```python
arr = np.arange(9).reshape(3, 3)
arr
```

### Создайте массив 4x3x2 со случайными значениями. Найти его минимум и максимум.

```python
arr = np.random.rand(4, 3, 2)
print("Массив:\n", arr)
print("Минимум:", arr.min())
print("Максимум:", arr.max())
```

### Создайте два двумерных массива размерами 6x4 и 4x3 и произведите их матричное умножение. 

```python
a = np.random.rand(6, 4)
b = np.random.rand(4, 3)
result = a @ b
print("Результат матричного умножения (6x3):\n", result)
```

### Создайте случайный двумерный массив 7x7, найти у него среднее и стандартное оклонение. Нормализуйте этот массив.

```python
arr = np.random.rand(7, 7)
mean = arr.mean()
std = arr.std()
normalized = (arr - mean) / std
print("Среднее:", mean)
print("Стандартное отклонение:", std)
print("Нормализованный массив:\n", normalized)
```

### Узнайте какой средний 'total_bill' в зависимости от 'day'

```python
tips.groupby("day", observed=True)["total_bill"].mean()
```

### Отберите строчки с 'total_bill' больше медианы и узнайте какой средний 'tip' в зависимости от 'sex'

```python
median_bill = tips["total_bill"].median()
above_median = tips[tips["total_bill"] > median_bill]
above_median.groupby("sex", observed=True)["tip"].mean()
```

### Преобразуйте признак 'smoker' в бинарный (0-No, 1-Yes)

```python
tips["smoker"] = tips["smoker"].map({"No": 0, "Yes": 1}).astype(int)
tips.head()
```

# III. Visualization

### Постройте гистограмму распределения признака 'total_bill'

```python
plt.figure(figsize=(8, 5))
sns.histplot(tips["total_bill"], bins=20, kde=True)
plt.title("Распределение total_bill")
plt.xlabel("total_bill")
plt.ylabel("Частота")
plt.show()
```

### Постройте scatterplot, представляющий взаимосвязь между признаками 'total_bill' и 'tip'

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=tips, x="total_bill", y="tip")
plt.title("Взаимосвязь total_bill и tip")
plt.show()
```

### Постройте pairplot

```python
sns.pairplot(tips)
plt.show()
```

### Постройте график взаимосвязи между признаками 'total_bill' и 'day'

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=tips, x="day", y="total_bill")
plt.title("total_bill в зависимости от дня недели")
plt.show()
```

### Постройте две гистограммы распределения признака 'tip' в зависимости от категорий 'time'

```python
g = sns.FacetGrid(tips, col="time", height=4, aspect=1.2)
g.map(sns.histplot, "tip", bins=15, kde=True)
g.fig.suptitle("Распределение tip по времени посещения", y=1.02)
plt.show()
```

### Постройте два графика scatterplot, представляющих взаимосвязь между признаками 'total_bill' и 'tip' один для Male, другой для Female и раскрасьте точки в зависимоти от признака 'smoker'

```python
g = sns.FacetGrid(tips, col="sex", height=4, aspect=1.2)
g.map_dataframe(sns.scatterplot, x="total_bill", y="tip", hue="smoker")
g.add_legend(title="smoker (0=No, 1=Yes)")
g.fig.suptitle("total_bill vs tip по полу и статусу курения", y=1.02)
plt.show()
```
