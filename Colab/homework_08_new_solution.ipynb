## Задание 4 — минимум суммы модулей отклонений

По рисунку точки имеют координаты

$$(-3,2),\quad(2,2),\quad(5,-3).$$

**Шаг 1.** Прямая проходит через начало координат, поэтому достаточно одного параметра — углового коэффициента $k$:

$$\widehat y=kx.$$

**Шаг 2.** Сумма вертикальных отклонений равна

$$
L(k)=|2-k(-3)|+|2-2k|+|-3-5k|
=|2+3k|+|2-2k|+|-3-5k|.
$$

Удобно вынести коэффициенты при модулях:

$$L(k)=3\left|k+\frac23\right|+2|k-1|+5\left|k+\frac35\right|.$$

**Шаг 3.** Это взвешенная сумма расстояний от $k$ до чисел $-\frac23$, $1$ и $-\frac35$ с весами $3$, $2$ и $5$. Минимум достигается во взвешенной медиане. Суммарный вес равен $10$, а точка $-\frac35$ имеет по обе стороны веса $3$ и $2$, не превосходящие $5$. Поэтому

$$k=-\frac35.$$

Проверим значение целевой функции:

$$
L\left(-\frac35\right)
=\left|2-\frac95\right|+\left|2+\frac65\right|+|-3+3|
=\frac15+\frac{16}{5}+0=\frac{17}{5}.
$$

**Ответ:** $\boxed{y=-\frac35x}$; минимальная сумма отклонений равна $\frac{17}{5}$.

## Задание 5 — минимум суммы квадратов отклонений

Используем ту же прямую $\widehat y=kx$.

**Шаг 1.** Единственный параметр — $k$.

**Шаг 2.** Целевая функция:

$$
Q(k)=(2+3k)^2+(2-2k)^2+(-3-5k)^2.
$$

Раскроем скобки:

$$Q(k)=38k^2+34k+17.$$

**Шаг 3.** Поскольку коэффициент при $k^2$ положителен, стационарная точка является точкой глобального минимума:

$$Q'(k)=76k+34=0
\quad\Longrightarrow\quad
k=-\frac{34}{76}=-\frac{17}{38}.$$

Также

$$Q''(k)=76>0,$$

что подтверждает минимум. Минимальное значение:

$$Q\left(-\frac{17}{38}\right)=\frac{357}{38}.$$

**Ответ:** $\boxed{y=-\frac{17}{38}x}$; минимальная сумма квадратов отклонений равна $\frac{357}{38}$.

```python
# Шаг 4 для заданий 4 и 5: изобразим точки и найденные прямые
import numpy as np
import matplotlib.pyplot as plt

x_points = np.array([-3, 2, 5])
y_points = np.array([2, 2, -3])
x_line = np.linspace(-5, 7, 300)

plt.figure(figsize=(9, 6))
plt.scatter(x_points, y_points, color="red", s=80, zorder=3, label="Исходные точки")
plt.plot(x_line, -3 / 5 * x_line, label=r"Минимум $\sum|e_i|$: $y=-\frac{3}{5}x$")
plt.plot(x_line, -17 / 38 * x_line, "--", label=r"Минимум $\sum e_i^2$: $y=-\frac{17}{38}x$")
plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)
plt.grid(alpha=0.3)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Оптимальные прямые, проходящие через (0, 0)")
plt.legend()
plt.show()
```

## Задание 6

Матрично-векторное уравнение

$$
\begin{pmatrix}1&2&4\\2&1&-1\\1&-1&1\end{pmatrix}
\begin{pmatrix}x_1\\x_2\\x_3\end{pmatrix}
=
\begin{pmatrix}3\\-9\\0\end{pmatrix}
$$

эквивалентно системе

$$
\begin{cases}
x_1+2x_2+4x_3=3,\\
2x_1+x_2-x_3=-9,\\
x_1-x_2+x_3=0.
\end{cases}
$$

Решим методом Гаусса. Запишем расширенную матрицу и выполним элементарные преобразования:

$$
\left(\begin{array}{ccc|c}
1&2&4&3\\
2&1&-1&-9\\
1&-1&1&0
\end{array}\right)
\xrightarrow{\substack{R_2\leftarrow R_2-2R_1\\R_3\leftarrow R_3-R_1}}
\left(\begin{array}{ccc|c}
1&2&4&3\\
0&-3&-9&-15\\
0&-3&-3&-3
\end{array}\right).
$$

Делим вторую строку на $-3$, затем прибавляем к третьей строке утроенную вторую:

$$
\left(\begin{array}{ccc|c}
1&2&4&3\\
0&1&3&5\\
0&0&6&12
\end{array}\right).
$$

Обратная подстановка дает

$$x_3=2,\qquad x_2+3\cdot2=5\Rightarrow x_2=-1,$$

$$x_1+2(-1)+4\cdot2=3\Rightarrow x_1=-3.$$

**Ответ:**

$$\boxed{x=\begin{pmatrix}-3\\-1\\2\end{pmatrix}}.$$

## Задание 7

По рисунку плотность задана прямой $f(x)=2x$ на отрезке $[0,1]$ и равна нулю вне этого отрезка:

$$
f(x)=
\begin{cases}
2x,&0\le x\le1,\\
0,&\text{иначе}.
\end{cases}
$$

Это действительно плотность, поскольку

$$\int_0^1 2x\,dx=\left.x^2\right|_0^1=1.$$

### 7.1. Вероятность попадания в $[0;0{,}5]$

Для непрерывного распределения включение границ не меняет вероятность:

$$
P(0\le X\le0{,}5)=\int_0^{0.5}2x\,dx
=\left.x^2\right|_0^{0.5}=0{,}25.
$$

### 7.2. Функция распределения (CDF)

По определению $F(t)=P(X\le t)$. Поэтому

$$
F(t)=
\begin{cases}
0,&t<0,\\
\displaystyle\int_0^t2x\,dx=t^2,&0\le t\le1,\\
1,&t>1.
\end{cases}
$$

## Задание 8 (дополнительное)

Теперь ищем прямую среди **всех** прямых, то есть разрешаем свободный член:

$$\widehat y=w_0+w_1x.$$

Для точек $(-3,2)$, $(2,2)$ и $(5,-3)$ матрица признаков и вектор ответов имеют вид

$$
X=\begin{pmatrix}1&-3\\1&2\\1&5\end{pmatrix},
\qquad
y=\begin{pmatrix}2\\2\\-3\end{pmatrix}.
$$

Вычислим необходимые произведения:

$$
X^TX=
\begin{pmatrix}1&1&1\\-3&2&5\end{pmatrix}
\begin{pmatrix}1&-3\\1&2\\1&5\end{pmatrix}
=\begin{pmatrix}3&4\\4&38\end{pmatrix},
$$

$$
X^Ty=
\begin{pmatrix}1&1&1\\-3&2&5\end{pmatrix}
\begin{pmatrix}2\\2\\-3\end{pmatrix}
=\begin{pmatrix}1\\-17\end{pmatrix}.
$$

Определитель $X^TX$ равен $3\cdot38-4\cdot4=98\ne0$, поэтому обратная матрица существует:

$$
(X^TX)^{-1}=\frac1{98}\begin{pmatrix}38&-4\\-4&3\end{pmatrix}.
$$

По формуле МНК

$$
\begin{aligned}
w_{\min}
&=(X^TX)^{-1}X^Ty\\
&=\frac1{98}\begin{pmatrix}38&-4\\-4&3\end{pmatrix}
\begin{pmatrix}1\\-17\end{pmatrix}
=\frac1{98}\begin{pmatrix}106\\-55\end{pmatrix}
=\begin{pmatrix}\frac{53}{49}\\-\frac{55}{98}\end{pmatrix}.
\end{aligned}
$$

**Ответ:** оптимальная прямая

$$\boxed{\widehat y=\frac{53}{49}-\frac{55}{98}x}.$$

```python
# Проверка матричного решения МНК и график для задания 8
import numpy as np
import matplotlib.pyplot as plt

X = np.array([[1, -3], [1, 2], [1, 5]], dtype=float)
y = np.array([2, 2, -3], dtype=float)
w = np.linalg.inv(X.T @ X) @ X.T @ y
print(f"w0 = {w[0]:.6f}, w1 = {w[1]:.6f}")

x_line = np.linspace(-5, 7, 300)
plt.figure(figsize=(8, 5))
plt.scatter(X[:, 1], y, color="red", s=80, zorder=3, label="Исходные точки")
plt.plot(x_line, w[0] + w[1] * x_line, label="Прямая МНК")
plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)
plt.grid(alpha=0.3)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Метод наименьших квадратов со свободным членом")
plt.legend()
plt.show()
```
